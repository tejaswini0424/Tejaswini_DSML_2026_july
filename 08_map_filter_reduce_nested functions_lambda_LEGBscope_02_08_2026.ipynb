```python
# Assignment 19: Find Maximum in a List

#Write a function named `find_list_maximum()`.

### Requirements

# Accept a list of numbers.
# Do not use `max()`., Return the largest list element.,Handle an empty list by returning an appropriate message.

def find_list_maximum(numbers):
    # Handle empty list
    if not numbers:
        return "List is empty"

    # Assume the first element is the maximum
    maximum = numbers[0]

    # Compare each element with maximum
    for number in numbers:
        if number > maximum:
            maximum = number

    return maximum


# Example
numbers = [10, 25, 7, 45, 18]

result = find_list_maximum(numbers)

print("List:", numbers)
print("Maximum:", result)
```

```python
## Assignment 34: Prime Number Function

#Write a function named `is_prime()`.

### Requirements

# Accept one integer.
# Return `True` when the number is prime.
# Return `False` otherwise., Use the function to print all prime numbers from 1 to 100.

### Hint,A prime number has exactly two factors.

def is_prime(number):
    if number<2:
        return False
    for i in range(2,int(number**0.5)+1):
        if number%i==0:
            return False
    return True
print("prime numbers from 1 to 100")
for number in range(1,101):
    if is_prime(number):
        print(number,end=" ")
print()
```

```python
# Assignment 35: Fibonacci Function

#Write a function named `generate_fibonacci()`.

### Requirements

# Accept the number of terms.
# Return the complete Fibonacci series as a list.
#Handle zero and negative term counts, Do not use recursion in this assignment.

### Example Input, Enter number of terms: 7

### Expected Output,[0, 1, 1, 2, 3, 5, 8]

def generate_fibonacci(n):
    if n<=0:
        return "invalid number of terms"
    fibonacci_series=[0,1]
    while len(fibonacci_series)<n:
        next_number=fibonacci_series[-1]+fibonacci_series[-2]
        fibonacci_series.append(next_number)
    return fibonacci_series
print(generate_fibonacci(7))
print()

```

```python
## Assignment 56: Mask Email Addresses,Given a list of email addresses, mask the username portion.

### Example, bharath@gmail.com → b******@gmail.com

### Requirements

# Use `map()`., Keep the first character of the username., Replace remaining username characters with `*`.
# Keep the domain unchanged., Create a separate named function for masking.


def mask_email(email):
    username,domain=email.split("@")

    if len(username) <= 1:
        masked_username = username
    else:
        masked_username = username[0] + "*" * (len(username) - 1)

    return masked_username + "@" + domain

emails = [
    "bharath@gmail.com",
    "bablu5@gmail.com",
    "teju7@gmail.com"
]

masked_emails = list(map(mask_email, emails))
print("Original Emails:", emails)
print("Masked Emails:", masked_emails)
```

```python
## Assignment 69: Filter Valid Email Addresses

#Given a list of email strings:

### Requirements

# Use `filter()`., A basic valid email should:
  # Contain exactly one `@`., Have characters before `@`.
  # Contain a dot after `@`., Return only emails satisfying these conditions.

### Note,A complete regular expression is not required.

def is_valid_email(email):
    if "@" not in email:
        return False
    if not email.split("@")[0]:
        return False
    username,domain=email.split("@")
    if "." not in domain:
        return False
    if username == "" or domain == "":
        return False
    if domain.startswith(".") or domain.endswith("."):
        return False
    return True
emails=input("enter emails separated by spaces:").split()
valid_emails=list(filter(is_valid_email,emails))
print("original emails:",emails)
print("valid emails:",valid_emails)
print()
```

```python
## Assignment 92: Closure-Based Bank Balance

#Write a function named `create_bank_account()`.

### Requirements

# Store the balance in the outer function.
# Define inner functions for:
# Deposit
# Withdrawal
# Balance enquiry, Use `nonlocal` to update the balance.
# Reject withdrawals greater than the available balance., Return all three inner functions.

### Hint

#Return multiple functions: return deposit, withdraw, check_balance



def create_bank_account():
    # Balance stored in the outer function
    balance = 0

    # Deposit function
    def deposit(amount):
        nonlocal balance
        balance += amount
        return balance

    # Withdrawal function
    def withdraw(amount):
        nonlocal balance

        if amount > balance:
            return "Insufficient balance"

        balance -= amount
        return balance

    # Balance enquiry function
    def check_balance():
        return balance

    # Return all three inner functions
    return deposit, withdraw, check_balance


# Create a bank account
deposit, withdraw, check_balance = create_bank_account()

# Deposit money
print("After deposit:", deposit(1000))

# Withdraw money
print("After withdrawal:", withdraw(500))

# Check current balance
print("Current balance:", check_balance())

# Try to withdraw more than available balance
print("Withdrawal:", withdraw(1000))

# Check balance again
print("Current balance:", check_balance())
```

```python

## Assignment 93: Function Decorator

#Create a decorator named `execution_logger()`.

### Requirements

# The decorator should print a message before the original function executes.
# It should print another message after execution.
# Apply it to an addition function., Preserve the original function's return value.

### Expected Pattern,Starting function... Result: 30,Function completed.

def execution_logger(func):
    def wrapper(*args,**kwargs):
        print("starting function...")
        result=func(*args,**kwargs)
        print("function completed")
        return result
    return wrapper

@execution_logger
def add(a,b):
    return a+b
result=add(10,20)
print("result:",result)
print()
```

```python
## Assignment 99: LEGB Name Resolution

#Create variables with the same name at multiple scope levels.

### Requirements

#Use the variable name `value` in:

# Global scope, Enclosing function scope, Local function scope

#print the variable from different locations.

### Tasks

#1. Print the global value.
#2. Print the enclosing value from the outer function.
#3. Print the local value from the inner function.
#4. Remove the local declaration and observe the result.
#5. Remove the enclosing declaration and observe the result.
#6. Explain the LEGB lookup order in comments.

### LEGB Order; L → Local, E → Enclosing, G → Global, B → Built-in






# G → Global scope
value = "Global value"


def outer_function():
    # E → Enclosing function scope
    value = "Enclosing value"

    print("From outer function:", value)

    def inner_function():
        # L → Local scope
        value = "Local value"

        print("From inner function:", value)

    inner_function()


# 1. Print the global value
print("1. Global value:", value)

# 2. Print the enclosing value from the outer function
outer_function()


# ---------------------------------------------------
# 3. Local value
# ---------------------------------------------------

def outer_function_2():
    value = "Enclosing value"

    def inner_function():
        # Local value exists, so Python uses Local scope first
        value = "Local value"
        print("2. Local value:", value)

    inner_function()


outer_function_2()


# ---------------------------------------------------
# 4. Remove the local declaration
# ---------------------------------------------------

def outer_function_3():
    value = "Enclosing value"

    def inner_function():
        # No local 'value'
        # Python searches the Enclosing scope
        print("3. Without local declaration:", value)

    inner_function()


outer_function_3()


# ---------------------------------------------------
# 5. Remove the enclosing declaration
# ---------------------------------------------------

def outer_function_4():

    def inner_function():
        # No Local value
        # No Enclosing value
        # Python searches the Global scope
        print("4. Without local and enclosing:", value)

    inner_function()


outer_function_4()


# ---------------------------------------------------
# 6. LEGB Lookup Order
# ---------------------------------------------------

# LEGB means:
#
# L → Local
#     Python first searches inside the current function.
#
# E → Enclosing
#     If the name is not found locally, Python searches
#     the surrounding/enclosing function.
#
# G → Global
#     If it is not found in Local or Enclosing scope,
#     Python searches the global scope.
#
# B → Built-in
#     Finally, Python searches built-in names such as
#     len(), print(), sum(), max(), etc.
#
# Lookup order:
#
# Local → Enclosing → Global → Built-in
```

```python
#assignment 100



from functools import reduce

# ============================================================
# Assignment 100: Employee Performance Analysis System
# ============================================================

# ------------------------------------------------------------
# Global variable - used for LEGB demonstration
# ------------------------------------------------------------
company_name = "Edukron"


# ------------------------------------------------------------
# Sample Employee Data
# ------------------------------------------------------------
employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]


# ============================================================
# VALIDATION FUNCTIONS
# ============================================================

def validate_employees(employee_list):
    """
    Validate salary and performance-score values.
    Returns True if all values are valid.
    """

    if not employee_list:
        return False

    for employee in employee_list:

        salary = employee.get("salary")
        score = employee.get("performance_score")

        # Salary must be a number and cannot be negative
        if not isinstance(salary, (int, float)) or salary < 0:
            print(f"Invalid salary for {employee.get('name', 'Unknown')}.")
            return False

        # Performance score must be between 0 and 100
        if (
            not isinstance(score, (int, float))
            or score < 0
            or score > 100
        ):
            print(
                f"Invalid performance score for "
                f"{employee.get('name', 'Unknown')}."
            )
            return False

    return True


# ============================================================
# PART 1: DISPLAY EMPLOYEES
# ============================================================

def display_employees(employee_list):
    """Display all employee details."""

    if not employee_list:
        print("\nNo employees available.")
        return

    print("\n========== EMPLOYEE DETAILS ==========")

    for employee in employee_list:
        print(f"ID                 : {employee['id']}")
        print(f"Name               : {employee['name']}")
        print(f"Department         : {employee['department']}")
        print(f"Salary             : ₹{employee['salary']:,.2f}")
        print(f"Performance Score  : {employee['performance_score']}")
        print(f"Experience         : {employee['experience']} years")
        print("-" * 40)


# ============================================================
# PART 2: EXTRACT EMPLOYEE NAMES USING map()
# ============================================================

def get_employee_names(employee_list):
    """Extract employee names using map()."""

    if not employee_list:
        return []

    names = list(
        map(lambda employee: employee["name"], employee_list)
    )

    return names


# ============================================================
# PART 3: FILTER HIGH PERFORMERS
# ============================================================

def get_high_performers(employee_list):
    """Return employees with performance score >= 80."""

    if not employee_list:
        return []

    high_performers = list(
        filter(
            lambda employee: employee["performance_score"] >= 80,
            employee_list
        )
    )

    return high_performers


# ============================================================
# PART 4: FILTER BY DEPARTMENT
# ============================================================

def get_department_employees(employee_list, department):
    """
    Return employees belonging to the requested department
    using filter().
    """

    if not employee_list:
        return []

    department = department.strip().lower()

    matching_employees = list(
        filter(
            lambda employee:
                employee["department"].lower() == department,
            employee_list
        )
    )

    return matching_employees


# ============================================================
# PART 5: CALCULATE REVISED SALARIES USING map()
# ============================================================

def calculate_revised_salaries(employee_list):
    """
    Calculate revised salaries using map().

    Original employee dictionaries are not modified.
    """

    if not employee_list:
        return []

    def calculate_salary(employee):

        score = employee["performance_score"]
        salary = employee["salary"]

        if score >= 90:
            increase = 0.15
        elif score >= 80:
            increase = 0.10
        elif score >= 70:
            increase = 0.05
        else:
            increase = 0

        revised_salary = salary * (1 + increase)

        # Create a new dictionary instead of modifying original
        new_employee = employee.copy()

        new_employee["revised_salary"] = revised_salary
        new_employee["increase_percentage"] = increase * 100

        return new_employee

    revised_employees = list(
        map(calculate_salary, employee_list)
    )

    return revised_employees


# ============================================================
# PART 6: TOTAL SALARY EXPENSE USING reduce()
# ============================================================

def calculate_salary_expense(employee_list):
    """
    Calculate:
    1. Current total salary
    2. Revised total salary
    3. Additional salary expense
    """

    if not employee_list:
        return 0, 0, 0

    revised_employees = calculate_revised_salaries(employee_list)

    current_total = reduce(
        lambda total, employee: total + employee["salary"],
        employee_list,
        0
    )

    revised_total = reduce(
        lambda total, employee:
            total + employee["revised_salary"],
        revised_employees,
        0
    )

    additional_expense = revised_total - current_total

    return current_total, revised_total, additional_expense


# ============================================================
# PART 7: HIGHEST-PAID EMPLOYEE USING reduce()
# ============================================================

def find_highest_paid(employee_list):
    """Find highest-paid employee without using max()."""

    if not employee_list:
        return None

    highest_paid = reduce(
        lambda employee1, employee2:
            employee1
            if employee1["salary"] >= employee2["salary"]
            else employee2,
        employee_list
    )

    return highest_paid


# ============================================================
# PART 8: SORT EMPLOYEES USING sorted() AND lambda
# ============================================================

def sort_employees(employee_list):
    """Display employees sorted using lambda expressions."""

    if not employee_list:
        print("\nNo employees available.")
        return

    print("\n========== SORTING OPTIONS ==========")
    print("1. Sort by Name")
    print("2. Sort by Salary")
    print("3. Sort by Performance Score")
    print("4. Sort by Experience")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        sorted_list = sorted(
            employee_list,
            key=lambda employee: employee["name"]
        )

        print("\nEmployees sorted by Name:")

    elif choice == "2":
        sorted_list = sorted(
            employee_list,
            key=lambda employee: employee["salary"]
        )

        print("\nEmployees sorted by Salary:")

    elif choice == "3":
        sorted_list = sorted(
            employee_list,
            key=lambda employee: employee["performance_score"],
            reverse=True
        )

        print("\nEmployees sorted by Performance Score:")

    elif choice == "4":
        sorted_list = sorted(
            employee_list,
            key=lambda employee: employee["experience"],
            reverse=True
        )

        print("\nEmployees sorted by Experience:")

    else:
        print("\nInvalid sorting choice.")
        return

    print("-" * 50)

    for employee in sorted_list:
        print(
            f"{employee['name']:10} | "
            f"Salary: ₹{employee['salary']:,.0f} | "
            f"Score: {employee['performance_score']} | "
            f"Experience: {employee['experience']} years"
        )


# ============================================================
# PART 9: CLOSURE-BASED BONUS CALCULATOR
# ============================================================

def create_bonus_calculator(bonus_percentage):
    """
    Outer function stores bonus_percentage.

    The inner function remembers bonus_percentage even
    after the outer function has finished.
    """

    def calculate_bonus(salary):
        return salary * bonus_percentage / 100

    return calculate_bonus


# Create three separate closure functions
bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)


def demonstrate_bonus_calculators(employee_list):
    """Demonstrate the three closure-based calculators."""

    if not employee_list:
        print("\nNo employees available.")
        return

    print("\n========== BONUS CALCULATORS ==========")

    for employee in employee_list:

        salary = employee["salary"]

        print(f"\nEmployee: {employee['name']}")
        print(f"Salary: ₹{salary:,.2f}")
        print(f"5% Bonus : ₹{bonus_5(salary):,.2f}")
        print(f"10% Bonus: ₹{bonus_10(salary):,.2f}")
        print(f"15% Bonus: ₹{bonus_15(salary):,.2f}")


# ============================================================
# PART 10 & 11:
# NESTED REPORT GENERATOR + LEGB DEMONSTRATION
# ============================================================

def generate_department_report(employee_list, department):
    """
    Generate a department report using nested functions.

    LEGB:
    L = Local scope
    E = Enclosing scope
    G = Global scope
    B = Built-in scope
    """

    # --------------------------------------------------------
    # Enclosing scope variable
    # This variable belongs to generate_department_report().
    # --------------------------------------------------------
    department_name = department.strip()

    # --------------------------------------------------------
    # Global scope
    # company_name is not local to this function.
    # Python searches the Global scope and finds it there.
    # --------------------------------------------------------

    # Local scope
    # report_title exists only inside this function.
    report_title = f"{company_name} - Department Report"

    # Filter employees for this department
    department_employees = get_department_employees(
        employee_list,
        department_name
    )

    # --------------------------------------------------------
    # Nested Function 1
    # --------------------------------------------------------
    def count_employees():
        # department_employees is found in the enclosing scope.
        return len(department_employees)

    # --------------------------------------------------------
    # Nested Function 2
    # --------------------------------------------------------
    def calculate_total_salary():
        if not department_employees:
            return 0

        return reduce(
            lambda total, employee:
                total + employee["salary"],
            department_employees,
            0
        )

    # --------------------------------------------------------
    # Nested Function 3
    # --------------------------------------------------------
    def calculate_average_salary():
        count = count_employees()

        if count == 0:
            return 0

        return calculate_total_salary() / count

    # --------------------------------------------------------
    # Nested Function 4
    # --------------------------------------------------------
    def calculate_average_performance():
        count = count_employees()

        if count == 0:
            return 0

        total_score = reduce(
            lambda total, employee:
                total + employee["performance_score"],
            department_employees,
            0
        )

        return total_score / count

    # --------------------------------------------------------
    # LEGB demonstration:
    #
    # L - report_title is found in the local scope.
    #
    # E - department_name is found in the enclosing scope
    #     when nested functions need it.
    #
    # G - company_name is found in the global scope.
    #
    # B - len(), print(), round() etc. come from built-in scope.
    # --------------------------------------------------------

    report = {
        "company": company_name,
        "report_title": report_title,
        "department": department_name,
        "employee_count": count_employees(),
        "total_salary": calculate_total_salary(),
        "average_salary": calculate_average_salary(),
        "average_performance":
            calculate_average_performance()
    }

    return report


# ============================================================
# DISPLAY DEPARTMENT REPORT
# ============================================================

def display_department_report(employee_list):
    """Ask for department and display its report."""

    if not employee_list:
        print("\nNo employees available.")
        return

    department = input(
        "Enter department name (IT/HR/Finance): "
    ).strip()

    report = generate_department_report(
        employee_list,
        department
    )

    if report["employee_count"] == 0:
        print(
            f"\nDepartment '{department}' does not exist "
            f"or has no employees."
        )
        return

    print("\n========== DEPARTMENT REPORT ==========")
    print(f"Company              : {report['company']}")
    print(f"Report Title         : {report['report_title']}")
    print(f"Department           : {report['department']}")
    print(f"Number of Employees  : {report['employee_count']}")
    print(
        f"Total Salary         : "
        f"₹{report['total_salary']:,.2f}"
    )
    print(
        f"Average Salary       : "
        f"₹{report['average_salary']:,.2f}"
    )
    print(
        f"Average Performance  : "
        f"{report['average_performance']:.2f}"
    )


# ============================================================
# DISPLAY HIGH PERFORMERS
# ============================================================

def display_high_performers(employee_list):

    high_performers = get_high_performers(employee_list)

    if not high_performers:
        print("\nNo high-performing employees found.")
        return

    print("\n========== HIGH PERFORMERS ==========")

    for employee in high_performers:
        print(
            f"{employee['name']} - "
            f"Performance: {employee['performance_score']}"
        )


# ============================================================
# DISPLAY REVISED SALARIES
# ============================================================

def display_revised_salaries(employee_list):

    revised_employees = calculate_revised_salaries(
        employee_list
    )

    if not revised_employees:
        print("\nNo employees available.")
        return

    print("\n========== REVISED SALARIES ==========")

    for employee in revised_employees:
        print(
            f"{employee['name']:10} | "
            f"Old Salary: ₹{employee['salary']:,.2f} | "
            f"Increase: {employee['increase_percentage']:.0f}% | "
            f"New Salary: ₹{employee['revised_salary']:,.2f}"
        )


# ============================================================
# DISPLAY SALARY EXPENSE
# ============================================================

def display_salary_expense(employee_list):

    current, revised, additional = calculate_salary_expense(
        employee_list
    )

    if not employee_list:
        print("\nNo employees available.")
        return

    print("\n========== SALARY EXPENSE ==========")
    print(f"Current Total Salary     : ₹{current:,.2f}")
    print(f"Revised Total Salary     : ₹{revised:,.2f}")
    print(f"Additional Salary Expense: ₹{additional:,.2f}")


# ============================================================
# DISPLAY HIGHEST PAID EMPLOYEE
# ============================================================

def display_highest_paid(employee_list):

    employee = find_highest_paid(employee_list)

    if employee is None:
        print("\nNo employees available.")
        return

    print("\n========== HIGHEST-PAID EMPLOYEE ==========")
    print(f"Name       : {employee['name']}")
    print(f"Department : {employee['department']}")
    print(f"Salary     : ₹{employee['salary']:,.2f}")


# ============================================================
# DISPLAY EMPLOYEE NAMES
# ============================================================

def display_names(employee_list):

    names = get_employee_names(employee_list)

    if not names:
        print("\nNo employees available.")
        return

    print("\n========== EMPLOYEE NAMES ==========")

    for name in names:
        print(name)


# ============================================================
# SEARCH BY DEPARTMENT
# ============================================================

def search_department(employee_list):

    if not employee_list:
        print("\nNo employees available.")
        return

    department = input(
        "Enter department name: "
    ).strip()

    matching_employees = get_department_employees(
        employee_list,
        department
    )

    if not matching_employees:
        print(
            f"\nNo employees found in department "
            f"'{department}'."
        )
        return

    print(
        f"\nEmployees in {department.title()} department:"
    )

    for employee in matching_employees:
        print(
            f"{employee['name']} - "
            f"Salary: ₹{employee['salary']:,.2f}"
        )


# ============================================================
# MENU
# ============================================================

def display_menu():

    print("\n")
    print("=" * 55)
    print("       EMPLOYEE PERFORMANCE ANALYSIS SYSTEM")
    print("=" * 55)
    print("1. Display all employees")
    print("2. Display employee names")
    print("3. Display high performers")
    print("4. Search by department")
    print("5. Calculate revised salaries")
    print("6. Calculate total salary expense")
    print("7. Find highest-paid employee")
    print("8. Sort employees")
    print("9. Generate department report")
    print("10. Exit")
    print("=" * 55)


# ============================================================
# MAIN PROGRAM
# ============================================================

def main():

    # Validate the initial employee data
    if not validate_employees(employees):
        print("Employee data contains invalid values.")
        return

    while True:

        display_menu()

        choice = input("Enter your choice: ").strip()

        if choice == "1":

            display_employees(employees)

        elif choice == "2":

            display_names(employees)

        elif choice == "3":

            display_high_performers(employees)

        elif choice == "4":

            search_department(employees)

        elif choice == "5":

            display_revised_salaries(employees)

        elif choice == "6":

            display_salary_expense(employees)

        elif choice == "7":

            display_highest_paid(employees)

        elif choice == "8":

            sort_employees(employees)

        elif choice == "9":

            display_department_report(employees)

        elif choice == "10":

            print("\nThank you for using the system!")
            print("Program ended.")
            break

        else:

            print(
                "\nInvalid menu choice. "
                "Please enter a number from 1 to 10."
            )


# ============================================================
# PROGRAM START
# ============================================================

if __name__ == "__main__":
    main()
```
