````python
import numpy as np

# ============================================================
# SMARTMART RETAIL ANALYTICS PROJECT
# ============================================================

stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])


# ============================================================
# PROFIT
# ============================================================

profit_data = sales_data - cost_data


# ============================================================
# EXAMPLE TARGETS
# Replace with your assignment's exact targets if provided
# ============================================================

sales_targets = np.array([
    800000,
    400000,
    300000,
    250000,
    180000,
    220000
])


# ============================================================
# EXAMPLE UNITS SOLD
# Replace with your assignment's exact units data if provided
# ============================================================

units_sold = np.array([
    [850, 420, 310, 275, 190, 225],
    [780, 460, 295, 310, 210, 240],
    [920, 390, 340, 260, 185, 215],
    [1050, 510, 375, 345, 250, 290],
    [690, 405, 280, 225, 175, 205]
])


# ============================================================
# EXAMPLE INVENTORY DATA
# ============================================================

closing_stock = np.array([
    [80, 150, 120, 90, 110, 95],
    [75, 140, 130, 85, 105, 90],
    [60, 125, 115, 70, 95, 85],
    [55, 100, 90, 65, 80, 75],
    [70, 135, 125, 80, 100, 90]
])

reorder_level = 100


# ============================================================
# COMPANY SUMMARY
# ============================================================

total_company_sales = np.sum(sales_data)
total_company_cost = np.sum(cost_data)
total_company_profit = np.sum(profit_data)

overall_margin = (
    total_company_profit
    / total_company_sales
) * 100

average_transaction_value = np.mean(sales_data)
highest_sales = np.max(sales_data)
lowest_sales = np.min(sales_data)


print("\n")
print("SMARTMART RETAIL PERFORMANCE REPORT")
print("===================================")

print("\nCOMPANY SUMMARY")
print("---------------")
print("Total Sales:", total_company_sales)
print("Total Cost:", total_company_cost)
print("Total Profit:", total_company_profit)
print("Overall Profit Margin:", round(overall_margin, 2), "%")
print("Average Transaction Value:", round(average_transaction_value, 2))
print("Highest Individual Sales Value:", highest_sales)
print("Lowest Individual Sales Value:", lowest_sales)


# ============================================================
# STORE SUMMARY
# ============================================================

store_sales = np.sum(sales_data, axis=1)
store_cost = np.sum(cost_data, axis=1)
store_profit = np.sum(profit_data, axis=1)

store_margin = (
    store_profit
    / store_sales
) * 100

target_achieved = sales_data >= sales_targets

number_targets_achieved = np.sum(
    target_achieved,
    axis=1
)

total_units_store = np.sum(
    units_sold,
    axis=1
)

# Expansion criteria:
# Sales > 2,500,000
# Profit > 700,000
# Margin > 25%

eligible_for_expansion = (
    (store_sales > 2500000)
    &
    (store_profit > 700000)
    &
    (store_margin > 25)
)


print("\nSTORE SUMMARY")
print("-------------")

for i in range(len(stores)):
    print("\nStore:", stores[i])
    print("Total Sales:", store_sales[i])
    print("Total Cost:", store_cost[i])
    print("Total Profit:", store_profit[i])
    print("Profit Margin:", round(store_margin[i], 2), "%")
    print(
        "Category Targets Achieved:",
        number_targets_achieved[i]
    )
    print("Total Units Sold:", total_units_store[i])
    print(
        "Expansion Eligibility:",
        "Eligible" if eligible_for_expansion[i]
        else "Not Eligible"
    )


# ============================================================
# CATEGORY SUMMARY
# ============================================================

category_sales = np.sum(sales_data, axis=0)
category_cost = np.sum(cost_data, axis=0)
category_profit = np.sum(profit_data, axis=0)

category_margin = (
    category_profit
    / category_sales
) * 100

category_average_sales = np.mean(
    sales_data,
    axis=0
)

category_targets_achieved = np.sum(
    sales_data >= sales_targets,
    axis=0
)

category_units = np.sum(
    units_sold,
    axis=0
)


print("\nCATEGORY SUMMARY")
print("----------------")

for i in range(len(categories)):
    print("\nCategory:", categories[i])
    print("Total Sales:", category_sales[i])
    print("Total Cost:", category_cost[i])
    print("Total Profit:", category_profit[i])
    print("Profit Margin:", round(category_margin[i], 2), "%")
    print("Average Sales:", round(category_average_sales[i], 2))
    print(
        "Stores Achieving Target:",
        category_targets_achieved[i]
    )
    print("Total Units Sold:", category_units[i])


# ============================================================
# MANAGEMENT RECOMMENDATIONS
# ============================================================

# Best and lowest-performing stores based on profit
best_store_index = np.argmax(store_profit)
lowest_store_index = np.argmin(store_profit)

# Most and least profitable categories
most_profitable_category_index = np.argmax(category_profit)
lowest_profit_category_index = np.argmin(category_profit)

# Expansion stores
expansion_stores = stores[eligible_for_expansion]


print("\nMANAGEMENT RECOMMENDATIONS")
print("==========================")

print("\nBEST-PERFORMING STORE")
print("---------------------")
print("Store:", stores[best_store_index])
print("Sales:", store_sales[best_store_index])
print("Cost:", store_cost[best_store_index])
print("Profit:", store_profit[best_store_index])
print(
    "Profit Margin:",
    round(store_margin[best_store_index], 2),
    "%"
)


print("\nLOWEST-PERFORMING STORE")
print("-----------------------")
print("Store:", stores[lowest_store_index])
print("Sales:", store_sales[lowest_store_index])
print("Cost:", store_cost[lowest_store_index])
print("Profit:", store_profit[lowest_store_index])
print(
    "Profit Margin:",
    round(store_margin[lowest_store_index], 2),
    "%"
)


print("\nMOST PROFITABLE CATEGORY")
print("------------------------")
print("Category:", categories[most_profitable_category_index])
print(
    "Sales:",
    category_sales[most_profitable_category_index]
)
print(
    "Cost:",
    category_cost[most_profitable_category_index]
)
print(
    "Profit:",
    category_profit[most_profitable_category_index]
)
print(
    "Profit Margin:",
    round(
        category_margin[most_profitable_category_index],
        2
    ),
    "%"
)


print("\nLOWEST-PROFIT CATEGORY")
print("----------------------")
print("Category:", categories[lowest_profit_category_index])
print(
    "Profit:",
    category_profit[lowest_profit_category_index]
)


print("\nEXPANSION RECOMMENDATIONS")
print("-------------------------")

if len(expansion_stores) > 0:
    for store in expansion_stores:
        print(store)
else:
    print("No stores are eligible.")


# ============================================================
# COST OPTIMIZATION
# ============================================================

# Find combinations with profit margin below 22%
combination_margin = (
    profit_data
    / sales_data
) * 100

low_margin = combination_margin < 22

print("\nCOST OPTIMIZATION")
print("-----------------")

found = False

for i, j in zip(*np.where(low_margin)):
    found = True
    print(
        "Store:",
        stores[i],
        "| Category:",
        categories[j],
        "| Cost:",
        cost_data[i, j],
        "| Profit Margin:",
        round(combination_margin[i, j], 2),
        "%"
    )

if not found:
    print("No combinations require cost optimization.")


# ============================================================
# INVENTORY REORDER
# ============================================================

reorder_required = closing_stock < reorder_level

print("\nINVENTORY REORDER")
print("-----------------")

found = False

for i, j in zip(*np.where(reorder_required)):
    found = True

    reorder_quantity = (
        reorder_level - closing_stock[i, j]
    )

    print(
        "Store:",
        stores[i],
        "| Category:",
        categories[j],
        "| Closing Stock:",
        closing_stock[i, j],
        "| Reorder Quantity:",
        reorder_quantity
    )

if not found:
    print("No inventory reorder required.")
````
