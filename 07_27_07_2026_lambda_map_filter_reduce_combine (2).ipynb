```python
# Using filter and reduce
# Extract palindrome words

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]

from functools import reduce
palindrome_words = list(
    filter(lambda word: word == word[::-1], words)
)

result = reduce(
    lambda x, y: x + ", " + y,
    palindrome_words
)

print("words =", words)
print("palindrome words =", palindrome_words)
print("result =", result)
```

```python
14
#using filter and reduce
#extract students who passed,find the topper among passed students
students=[
          {"name":"ravi","marks":92,"passed":True},
          {"name":"kiran","marks":68,"passed":True},
          {"name":"meena","marks":85,"passed":True},
          {"name":"ajay","marks":48,"passed":True},
          {"name":"sneha","marks":35,"passed":False},
          {"name":"ajay","marks":95,"passed":True}
]
from functools import reduce
result=reduce(lambda x,y:x if x["marks"]>y["marks"] else y,filter(lambda x:x["passed"],students))
print("students=",students)
print("result=",result)
```

```python
15
#using filter,map and reduce,extract students scoring more than 70,extract only their marks,calculate the average marks
students=[
          {"name":"ravi","marks":92},
          {"name":"kiran","marks":68},
          {"name":"meena","marks":85},
          {"name":"ajay","marks":48},
          {"name":"sneha","marks":79},
          {"name":"rahul","marks":95}
]
from functools import reduce
#filter students with marks > 70
passed_students=list(filter(lambda x:x["marks"]>70,students))
#extract only marks
marks=list(map(lambda x:x["marks"],passed_students))
#calculate average marks
result=reduce(lambda x,y:x+y,marks)/len(marks)
print("students=",students)
print("result=",result)
```

```python
16
#using fiter,map and reduce,
#extract products currently in stock,calculate each products inventory value using price*stock,calculate the total inventory value
products=[
          {"name":"laptop","price":55000,"stock":5},
          {"name":"mouse","price":600,"stock":25},
          {"name":"keyboard","price":1500,"stock":0},
          {"name":"monitor","price":18000,"stock":8},
          {"name":"printer","price":12000,"stock":0},
          {"name":"ssd","price":6500,"stock":12}
]
from functools import reduce
#extract products in stock
in_stock_products=list(filter(lambda x:x["stock"]>0,products))
#calculate inventory value
inventory_value=reduce(lambda x,y:x+y["price"]*y["stock"],in_stock_products,0)
print("products=",products)
print("inventory_value=",inventory_value)
```

```python
17
#using filter,map and reduce
#extract products belonging to the electronics category,apply a 20% discount,calculate the total discounted value
products=[
          {"name":"laptop","price":55000,"category":"electronics"},
          {"name":"chair","price":4500,"category":"furniture"},
          {"name":"monitor","price":18000,"category":"electronics"},
          {"name":"ssd","price":8500,"category":"furniture"},
          {"name":"sofa","price":25000,"category":"electronics"}

]
from functools import reduce

#filter electronic products
electronics_products=list(filter(lambda x:x["category"]=="electronics",products))
#apply 20% discount
discounted_products=list(map(lambda x:{"name":x["name"],"price":x["price"]*0.8},electronics_products))
#calculate total discounted value
total_discounted_value=reduce(lambda x,y:x+y["price"],discounted_products,0)
print("products=",products)
print("total_discounted_value=",total_discounted_value)
```

```python
18
#using filter and reduce,extract delivered orders,calculate the total delivered revenue
orders=[
          {"order_id":1001,"amount":1200,"status":"delivered"},
          {"order_id":1002,"amount":3500,"status":"pending"},
          {"order_id":1003,"amount":8500,"status":"delivered"},
          {"order_id":1004,"amount":4200,"status":"cancelled"},
          {"order_id":1005,"amount":7600,"status":"delivered"},
          {"order_id":1006,"amount":5100,"status":"pending"}
]
from functools import reduce
#extract delivered orders
delivered_orders=list(filter(lambda x:x["status"]=="delivered",orders))
#calculate total delivered revenue
total_delivered_revenue=reduce(lambda x,y:x+y["amount"],delivered_orders,0)
print("orders=",orders)
print("total_delivered_revenue=",total_delivered_revenue)
```

```python
19
#using filter,map and reduce,extract transactions whose amount is greater than 5000,add 18% gst to every selecte transaction,calculate the grand total after gst
transactions=[
          {"transaction_id":1,"amount":1200},
          {"transaction_id":2,"amount":5200},
          {"transaction_id":3,"amount":18000},
          {"transaction_id":4,"amount":25000},
          {"transaction_id":5,"amount":9600},
          {"transaction_id":6,"amount":7200}
]
from functools import reduce
#extract transactions with amount > 5000
high_value_transactions=list(filter(lambda x:x["amount"]>5000,transactions))
#add 18% gst to every transaction
gst_transactions=list(map(lambda x:{"transaction_id":x["transaction_id"],"amount":x["amount"]*1.18},high_value_transactions))
#calculate grand total after gst
grand_total=reduce(lambda x,y:x+y["amount"],gst_transactions,0)
print("transactions=",transactions)
print("grand_total=",grand_total)
```

```python
20
#using filter,map and reduce
#extract ative salespeople from the south region,calculate a 5% incentive for each selected salesperson
#create a list containing salesperson name,sales amount,and incentive
sales_records=[
    {"salesperson":"rahul","region":"north","sales_amount":45000,"active":True},
    {"salesperson":"anil","region":"south","sales_amount":65000,"active":True},
    {"salesperson":"sneha","region":"south","sales_amount":72000,"active":False},
    {"salesperson":"kiran","region":"north","sales_amount":82000,"active":True},
    {"salesperson":"meena","region":"east","sales_amount":55000,"active":True},
    {"salesperson":"arjun","region":"north","sales_amount":98000,"active":True}
]
from functools import reduce
#south region salespeople
south_region_salespeople=list(filter(lambda x:x["region"]=="south" and x["active"],sales_records))
#calculate 5% incentive for each selected salesperson
incentives=list(map(lambda x:{"salesperson":x["salesperson"],"sales_amount":x["sales_amount"]*0.05},south_region_salespeople))
#create a list containing salesperson name,sales amount and incentive
result=list(map(lambda x,y:{"salesperson":x["salesperson"],"sales_amount":x["sales_amount"],"incentive":y["sales_amount"]},sales_records,incentives))
print("sales_records=",sales_records)
print("result=",result)
```
